```python
import sys
sys.path.append("..")

from src.preprocess import load_raw, drop_missing_and_duplicates, add_base_price, remove_outliers

df = load_raw("../data/raw_data.csv")
df = drop_missing_and_duplicates(df)
df = add_base_price(df)
df = remove_outliers(df)
df.shape
```

## Conclusion
The correlation of `hour` and the weather variables with `base_price` is practically zero.All coefficients lie between -0.0005 and 0.0018 (|r| < 0.002), so each variable explains
less than 0.001% of the variance of the base price.
